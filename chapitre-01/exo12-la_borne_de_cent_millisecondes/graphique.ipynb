## Importation de la bibliothèqe

```python
import matplotlib.pyplot as plt
```

## Réalisation du graphique

```python
durees_ms = [10, 20, 50, 100, 200, 300, 500, 750, 1000]
erreurs_deg = [0.0297, 0.1174, 0.7100, 2.6927, 9.7245, 19.8655, 46.1993, 85.4326, 127.9264]

plt.plot(durees_ms, erreurs_deg, marker='o')
plt.xlabel("Durée extrapolée (ms)")
plt.ylabel("Erreur angulaire (deg)")
plt.title("Erreur d'extrapolation vs durée")
plt.axvline(100, color='red', linestyle='--', label='borne 100 ms')
plt.legend()
plt.grid(True)
plt.show()
```
